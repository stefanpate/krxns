```python
kcs = pd.read_csv("../data/interim/compounds.csv")
sources = pd.read_csv("../data/interim/default_sources.csv")
sources = sources['id'].tolist()
with open("../data/interim/mass_contributions.json", 'r') as f:
    mass_contributions = json.load(f)

edges, nodes = construct_reaction_network(
    mass_contributions=mass_contributions,
    compounds=kcs,
    sources=sources,
    pnmc_lb=0.33,
    rnmc_lb=0.33,
    tot_mass_lb=0.7
)

G = SuperMultiDiGraph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

data = nx.node_link_data(G)
with open('typical_graph.json', 'w') as f:
    json.dump(data, f)
```

```python
edges, nodes = construct_reaction_network(
    mass_contributions=mass_contributions,
    compounds=kcs,
    sources=sources,
    pnmc_lb=0.0,
    rnmc_lb=0.0,
    tot_mass_lb=0.1
)

G = SuperMultiDiGraph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

data = nx.node_link_data(G)
with open('largish_graph.json', 'w') as f:
    json.dump(data, f)
```
